```python
import pandas as pd

input_file = "5year.arff"

with open(input_file, "r", encoding="utf-8") as f:
    lines = f.readlines()

# Find @data
data_index = next(
    i for i, line in enumerate(lines)
    if line.strip().lower() == "@data"
)

rows = []

for line in lines[data_index + 1:]:
    line = line.strip()

    if not line or line.startswith("%"):
        continue

    values = [x.strip() for x in line.split(",")]

    if len(values) == 65:
        rows.append(values)

# Create DataFrame with temporary names
columns = [f"Attr{i}" for i in range(1, 65)] + ["class"]

arff_df = pd.DataFrame(rows, columns=columns)

# Convert ? to NaN
arff_df = arff_df.replace("?", pd.NA)

print("ARFF missing values:")
print(arff_df.isna().sum()[arff_df.isna().sum() > 0])
```

```python
print("Total missing values:", arff_df.isna().sum().sum())
```
